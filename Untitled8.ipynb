```python
import random
from datetime import datetime

print("="*55)
print("🤖 AI RULE-BASED CHATBOT")
print("="*55)

name = input("Enter your name: ").title()
print(f"\nHello {name}! 😊")
print("Type 'help' to see available commands.")
print("Type 'bye' to exit.\n")

jokes = [
    "😂 Why don't programmers like nature? It has too many bugs.",
    "😂 Python is my favorite because it doesn't bite.",
    "😂 AI won't replace humans... but humans using AI might."
]

quotes = [
    "💡 Success comes from consistency.",
    "💡 Believe in yourself and keep learning.",
    "💡 Small progress is still progress."
]

while True:

    user = input(f"\n{name}: ").lower()

    # Greeting
    if any(word in user for word in ["hi","hello","hey"]):
        print("Bot: Hello! 👋 Nice to meet you.")

    # Name
    elif "your name" in user:
        print("Bot: My name is SmartBot 🤖")

    # How are you
    elif "how are you" in user:
        print("Bot: I'm doing great! Thanks for asking 😊")

    # User name
    elif "my name" in user:
        print(f"Bot: Your name is {name} 😄")

    # Time
    elif "time" in user:
        current = datetime.now().strftime("%I:%M %p")
        print("Bot: Current Time:", current)

    # Date
    elif "date" in user:
        current = datetime.now().strftime("%d-%m-%Y")
        print("Bot: Today's Date:", current)

    # Joke
    elif "joke" in user:
        print("Bot:", random.choice(jokes))

    # Motivation
    elif "motivate" in user or "motivation" in user:
        print("Bot:", random.choice(quotes))

    # Calculator
    elif user.startswith("calc"):
        try:
            exp = user.replace("calc","")
            result = eval(exp)
            print("Bot: Answer =", result)
        except:
            print("Bot: Invalid expression.")

    # Help
    elif user=="help":
        print("""
Available Commands
--------------------------
Hi / Hello
How are you
Your name
My name
Time
Date
Joke
Motivate me
Calc 10+20
Calc 50/2
Bye
        """)

    # Goodbye
    elif user=="bye":
        print("Bot: Goodbye 👋 Have a great day!")
        break

    # Thanks
    elif "thank" in user:
        print("Bot: You're Welcome 😊")

    # Default
    else:
        print("Bot: Sorry 😅 I didn't understand that.")
```
